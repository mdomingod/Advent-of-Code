```python
input=[]
with open("input2d1s.txt", "r") as f:
    for l in f:
        input = l

print(input)
    
```

```python
ranges = [tuple(map(int, num_range.split("-"))) for num_range in input.split(",")]
print(ranges)
```

```python
invalids_ids_sum = 0
# ranges = [(3, 5), (6, 9), (2, 6), (12, 15)]
for num1, num2, in ranges:
    for num in range(num1, num2+1):
        num_str = str(num)
        digits = len(num_str)

        if digits % 2 == 0:
            if num_str[:int(digits/2)] == num_str[int(digits/2):]:

                invalids_ids_sum += num

print(f"the total invalids ids sum is", invalids_ids_sum)
            
```
