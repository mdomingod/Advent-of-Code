```python

def count_zero_hits(SP, dist):
    """
    Count how many clicks (including the final one) land the dial on 0
    when starting at SP (0..99) and rotating by signed distance dist.
    Positive dist = R (increase), negative dist = L (decrease).
    """
    m = abs(dist)
    if m == 0:
        return 0

    if dist > 0:  # Right / increasing
        threshold = (100 - SP) % 100
        if threshold == 0:
            # Starting at 0: first landing on 0 is at 100 clicks, not 0
            threshold = 100
        if m < threshold:
            return 0
        return 1 + (m - threshold) // 100
    else:  # Left / decreasing
        threshold = SP % 100
        if threshold == 0:
            # Starting at 0: first landing on 0 is at 100 clicks, not 0
            threshold = 100
        if m < threshold:
            return 0
        return 1 + (m - threshold) // 100


def solve_file(filename, start=50):
    SP = start
    total = 0
    with open(filename, "r") as f:
        for l in f:
            l = l.strip()
            if not l:
                continue
            dirc, k = l[0], int(l[1:])
            dist = k if dirc == 'R' else -k
            total += count_zero_hits(SP, dist)
            SP = (SP + dist) % 100  # advance the dial
    return total
```
